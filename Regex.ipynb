```python
import pandas as pd
import re
data = pd.read_csv("/content/students.csv")
# print(data)
```

```python
# 1.
def find_email(text):
    match = re.match(r'([^@]+)', str(text))
    if match:
        return match.group(0)
    return None

data['username'] = data['email'].apply(lambda x: find_email(x))
print("Extracting email from dataframe column:")
print(data[['email', 'username']])
```

```python
# 2.
def find_state(text):
    if re.match(r'^[CT]', str(text)):
        return text
    return None

data['f_s'] = data['hs_state'].apply(find_state)
f_s = data['f_s'].dropna()
print(f_s)
```

```python
# 3.
data['hs_zip'] = data['hs_zip'].astype(str)
reg_zip = r'^8'
data1 = data[data['hs_zip'].str.match(reg_zip)]
print(data1[['gender', 'hs_zip']])
```

```python
# 4.
reg_name = r'^J'
data2 = data[data['first_name'].str.match(reg_name, na=False)]
print("Candidates with first name starting with J:")
print(data2[['first_name']])
```

```python
# 5.
data['id'] = data['id'].astype(str)
data['id'] = data['id'].str.replace('1111', '', regex=False)
print("Updated IDs after removing '1111':")
print(data[['id']])
```
